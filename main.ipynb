## Bài 1:
**Main idea**: Nhập giá trị bán kính và sủ dụng công thức để tính ra diện tích pi*r^2

**Mã giả**:
        
        nhập r
        tính s qua pi*r^2

**Code**:
        using System;
        namespace Bai1;
        class HinhTron
        {
        static void Main(String[] args)
        {
                Console.WriteLine("Nhap ban kinh hinh tron:");
                double r = double.Parse(Console.ReadLine());
                Console.WriteLine($"Dien tich hinh tron la: {Math.PI * r * r:F1}");
                Console.WriteLine($"Chu vi P = {2*Math.PI*r:F1}");
                
        }
        }
